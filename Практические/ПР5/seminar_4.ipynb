## Нормализация данных

```python
data = np.loadtxt("ex1data2.txt", delimiter=",")
X = data[:, :2]
y = data[:, 2]
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("Скорость оборота трактора")
plt.ylabel("Число передач")
plt.title("Ненормализованные данные", fontsize=18)
plt.show()
```

```python
start_time = perf_counter()
X_normalized = h.norm_manual(X)
manual_time = perf_counter() - start_time

plt.scatter(X_normalized[:, 0], X_normalized[:, 1])
plt.xlabel("Скорость оборота трактора")
plt.ylabel("Число передач")
plt.title(f"Поэлементная нормировка", fontsize=18)
plt.show()
```

```python
start_time = perf_counter()
X_normalized = h.norm_loop(X)
loop_time = perf_counter() - start_time


plt.scatter(X_normalized[:, 0], X_normalized[:, 1])
plt.xlabel("Скорость оборота трактора")
plt.ylabel("Число передач")
plt.title("Нормировка с ипользованием векторных методов", fontsize=18)
plt.show()
```

```python
start_time = perf_counter()
X_normalized = h.norm_vector(X)
vector_time = perf_counter() - start_time

plt.scatter(X_normalized[:, 0], X_normalized[:, 1])
plt.xlabel("Скорость оборота трактора")
plt.ylabel("Число передач")
plt.title("Нормировка с матричными преобразованиями", fontsize=18)
plt.show()
```

```python
X_normalized = h.norm_minmax(X)

plt.scatter(X_normalized[:, 0], X_normalized[:, 1])
plt.xlabel("Скорость оборота трактора")
plt.ylabel("Число передач")
plt.title("Нормировка MinMax", fontsize=18)
plt.show()
```

```python
X_normalized = h.norm_max(X)

plt.scatter(X_normalized[:, 0], X_normalized[:, 1])
plt.xlabel("Скорость оборота трактора")
plt.ylabel("Число передач")
plt.title("Нормировка Max", fontsize=18)
plt.show()
```

```python
print(f"Время нормировки поэлементным подходом: {manual_time * 1000:.2f} мс")
print(f"Время нормировки векторными методами: {loop_time * 1000:.2f} мс")
print(f"Время нормировки матричными преобразованиями: {vector_time * 1000:.2f} мс")
```
