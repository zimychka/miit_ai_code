## Задание 1

Написать функцию, построить график, объяснить поведение, составить отчет, отчет защитить

```python
def log_base(x, base):
    return np.log(x) / np.log(base)

def m_function(x, a=1, d=2, y=3):
    res = (
        np.sin(x + 2 * y)**2 / np.abs(x - np.exp(a * x * y)) + 
        0.3 * np.cbrt(x - (a + 4.2) / (d * y)) + 
        log_base(a * x**2, 5) +
        np.abs(x) / x
    )
    return res
```

```python
xs = np.linspace(-20, 20, 10000)
ys = m_function(xs, a=1, d=2, y=3)

plt.figure(figsize=(12, 8))

plt.plot(xs, ys, linewidth=1.5)
plt.axvline(x=0, linestyle='--', color='red', alpha=0.5)

plt.grid(alpha=0.3)
```
