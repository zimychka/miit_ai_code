# Задание 2

```python
def F(x, y):
    term_1 = 2 * np.exp(-1 * ((x - 5)**2 + (y + 0.2)**2) / 1.5)
    term_2 = 0.8 * np.exp(-1 * ((2 * x + 3)**2 + (0.2 * y - 4)**2) / 5)
    term_3 = -4 * np.exp(-1 * ((x - 0.2)**2 + (y + 2)**2) / 2)
    return term_1 + term_2 + term_3
```

```python
val = 26
x_min, x_max, y_min, y_max = -val, val, -val, val
n_points = 1000

xs, ys = np.linspace(x_min, x_max, n_points), np.linspace(y_min, y_max, n_points)

X, Y = np.meshgrid(xs, ys)
Z = F(X, Y)
```

```python
fig, ax = plt.subplots(subplot_kw={"projection": "3d"}, figsize=(12, 8))
surf = ax.plot_surface(X, Y, Z, cmap="coolwarm")
ax.view_init(azim=-130)
ax.set_xlabel("X")
ax.set_ylabel("Y")
fig.colorbar(surf, ax=ax)
ax.set_title("3D график", fontsize=18)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 8))
contour = ax.contour(X, Y, Z, levels=10, cmap='coolwarm')
fig.colorbar(surf, ax=ax)
ax.clabel(contour, colors='black', fontsize=7)
ax.set_title("График линий уровня", fontsize=18)
ax.grid(alpha=0.3)
plt.show()
```
