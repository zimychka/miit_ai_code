```python
xs = np.linspace(-15, 15, 300)
special_xs = np.array([-15, -8, -3, 0, 3, 8, 15])
```

```python
plt.plot(xs, h.sigmoid(xs), label="Исходная функция")
plt.scatter(special_xs, h.sigmoid(special_xs), color="red")
plt.plot(xs, h.sigmoid_derivative(xs), label="Производная функции", linestyle="--")
plt.title("Сигмоида", size=18)
plt.legend()
plt.grid(alpha=0.4)
plt.show()
```

```python
xs = np.linspace(-5, 5, 300)
plt.plot(xs, h.tanh(xs), label="Исходная функция")
plt.plot(xs, h.tanh_derivative(xs), label="Производная функции", linestyle="--")
plt.title("Гиперболический тангенс", size=18)
plt.legend()
plt.grid(alpha=0.4)
plt.show()
```

```python
plt.figure(figsize=(10, 8))
lim = 2.5
xs = np.linspace(-lim, lim, 100)
xs1 = np.linspace(-lim, -0.01, 50)
xs2 = np.linspace(0.01, lim, 50)
plt.plot(xs, h.sinh(xs), label="sinh(x)")
plt.plot(xs, h.cosh(xs), label="cosh(x)", linestyle="--")
plt.plot(xs, h.tanh(xs), label="tanh(x)", linestyle="-.")
plt.plot(xs1, h.ctanh(xs1), label="ctanh(x)", color="tab:red", linestyle=":")
plt.plot(xs2, h.ctanh(xs2), color="tab:red", linestyle=":")
plt.legend()
plt.ylim(-2, 2)
plt.title("Гиперболические функции", fontsize=18)
plt.show()
```
